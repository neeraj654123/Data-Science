# Text file .txt

```python
## Writing to a text file using built-in open() function
```

```python
with open('example.txt', 'w') as  file:
    file.write('This is Example file of text.\n')
print('file created successfully')
```

```python
# The upper command will be overwritten by this command
with open('example_2.txt', 'w') as  file:
    for i in range(1,6):
        file.write(f'This is the {i} line.\n')
print('file updated successfully')
```

### Writing multiple lines

```python
lines = ['This is the 1st line.',
         'This is the 2nd line.',
         'This is the 3rd line.',
         'This is the 4th line.',
         'This is the 5th line.']
with open('example_multiline.txt', 'w') as  file:
    for li in lines:
        file.write(li+'\n')
print('File created with multiple lines')
```

## Reading a text file using built-in open() function

```python
# To open file in read mode then existance of file is important
with open('example_2.txt', 'r') as  file:
    data = file.read()
    print(data)
```

# TSV (Tab Separated File) .tsv

## Writing data to a TSV file

```python
import pandas as pd
data = {'Emp_ID' : [101,102,103,104],
        'Name' : ['Rahul Sharma','Priya Patel', 'Amit Singh', 'Neeta Rao'],
        'Department' : ['HR', 'Finance', "IT", 'Marketing'],
        'Salary (₹)' : [50000,65000,40000,60000],
        'Joining_date' : ['2019-05-19','2020-06-23','2020-12-04','2023-03-20'],
        'City' : ['Mumbai', 'Delhi', 'Benglore', 'Hyderabad']
       }
data
```

```python
df = pd.DataFrame(data)
df
```

```python
df.to_csv('tab_example.tsv', sep='\t', index=False)
print('File created')
```

## Reading TSV file

```python
df_read = pd.read_csv('tab_example.tsv',sep ='\t')

# Display the content
print(df_read)
```

# JSON (JavaScript Onject Notation) .json

## Creating a JSON with Multiplelines

```python
import json

data = {"employees":[
                {'name':'John', 'age':30, 'department':'Sales'},
                {'name':'Meera', 'age':25, 'department':'Marketing'},
                {'name':'Rahul', 'age':35, 'department':'HR'}
]}

# Writing the dictionary to a JSON file
with open('json_example.json', 'w') as file:
    json.dump(data, file , indent=4)
```

## Reading the JSON File

```python
with open('json_example.json', 'r') as file:
    data = json.load(file)

# Printing data in more readable format
print('Employee Information')
for em in data['employees']:
    print(f"Name: {em['name']} Age: {em['age']} Department: {em['department']}")
```
