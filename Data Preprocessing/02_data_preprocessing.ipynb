# Feature Scaling

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Example dataset
numeric_data = pd.DataFrame({
    'Salary':[30000,45000,60000,80000,120000],
    'AGE': [25,32,47,51,62]
})
numeric_data
```

### Standard Scaling (Z-score)

```python
std_scaler = StandardScaler()
scaled_std = std_scaler.fit_transform(numeric_data)
df = pd.DataFrame(scaled_std,columns=numeric_data.columns)
print('\n Standard Scaling :\n',df)
```

### MinMax Scaling (0-1)

```python
minmax_scaler = MinMaxScaler()
scaled_minmax = minmax_scaler.fit_transform(numeric_data)
print('\nMinMax  Scaling :\n',pd.DataFrame(scaled_minmax,columns=numeric_data.columns))
```

# Encoding Categorical Variables

## Converts text labels to numeric format for ML models

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder

# Sample data
cat_data = pd.DataFrame({
    'Department': ['HR','Sales','IT','HR','IT','Finance']
})
cat_data
```

### Label Encoding

```python
label_encoder = LabelEncoder()
cat_data['Dept_label']=label_encoder.fit_transform(cat_data.Department)
print("Label Encoded:\n",cat_data)
```

### OneHot Encoding

```python
cat_data_ohe = pd.get_dummies(cat_data.Department,prefix='Dept')
print('OneHot Encoded: \n',cat_data_ohe)
```
