### Declare feature vector and target variable

```python
x = df.drop(['class'],axis=1) # 1 means column
y = df['class']
```

```python
y.value_counts()
```

### Split data into separate training and test sets

```python
# split X and y into training and testing sets
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.33, random_state = 42)
```

```python
import category_encoders as ce
```

```python
# encode variables with ordinal encoding
encoder = ce.OrdinalEncoder(cols=['buying', 'maint', 'doors', 'persons',
                                  'lug_boot', 'safety'])
```

```python
x_train = encoder.fit_transform(x_train)
x_test = encoder.transform(x_test)
```

```python
x_train.dtypes
```

### Decision Tree Classifier with criterion gini index

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf_gini = DecisionTreeClassifier(criterion='gini', max_depth=4, random_state=0)
```

```python
clf_gini.fit(x_train, y_train)
```

### Predict the Test set results with criterion gini index

```python
y_pred_gini = clf_gini.predict(x_test)
```

```python
y_pred_gini[:10]
```

```python
from sklearn.metrics import accuracy_score
print('Model accuracy with  gini index: {0:0.4f}'. format(accuracy_score(y_test, y_pred_gini)))
```

### Compare the train-set and test-set accuracy

```python
y_pred_train_gini = clf_gini.predict(x_train)
y_pred_train_gini
```

```python
print('Training-set accuracy score: {0:0.4f}'. format(accuracy_score(y_train, y_pred_train_gini)))
```

### Check for overfitting and underfitting

```python
print('Training set score: {:.4f}'.format(clf_gini.score(x_train, y_train)))

print('Test set score: {:.4f}'.format(clf_gini.score(x_test, y_test)))
```
