# Descriptive stats

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create a Sample dataframe 
data = {
    'Income':[25000,28000,30000,29000,27000,100000,26000],
    'Car_Color':['Red','Red','Blue','Black','Red','Blue','Black']
}
df =pd.DataFrame(data)
df
```

## Mean

```python
mean_income = df.Income.mean()
print('Mean Income:',mean_income)
```

## Median

```python
median_income = df.Income.median()
print('Median Income:',median_income)
```

## Mode

```python
mode_color = df.Car_Color.mode()[0]
print('Mode Color:',mode_color)
```

## Standard Deviation & Variance

```python
std_income = df.Income.std()
Variance_income = df.Income.var()
print("Standard Deviation:",std_income)
print("Variance of Income:",Variance_income)
```

## Range

```python
range_income = df.Income.max() - df.Income.min() 
print('Income Range:', range_income)
```

## Percentiles & Quartiles

```python
q1 = df.Income.quantile(0.25)
q2 = df.Income.quantile(0.50)
q3 = df.Income.quantile(0.75)
print("Q1:",q1)
print("Q2:",q2)
print("Q3:",q3)
```

## Boxplot of Visual Summary

```python
plt.figure(figsize=(8,8))
sns.boxplot(y=df.Income)
plt.title('Box Plot of Income')
plt.show()
```

## Histogram of Distribution

```python
plt.figure(figsize=(8,4))
sns.histplot(df.Income,kde=True)
plt.title('Distribution of Income')
plt.show()
```
