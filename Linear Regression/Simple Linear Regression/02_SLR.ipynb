# Predicting Saled Based on Store Footfalls using Simple Linear Regression

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('General_Stores.csv')
print("Dataset Preview:")
data
```

```python
X = data[['Footfalls']]
Y = data['DailySales']
```

```python
type(X), type(Y)
```

## Visualise data and check relation Between X and Y

```python
plt.xlabel("Footfalls")
plt.ylabel("Daily Sales")
plt.scatter(X,Y,color='red',marker='o')
```

## Create and train Linear Regression Model

```python
model = LinearRegression()
model.fit(X,Y)
```

```python
# fit is a method for creating model or training data or creating relations between X and Y
```

## Print the Model Parameters

```python
print(f"Coefficient (slope): {model.coef_}")
print(f"Incercept: {model.intercept_}")
```

## Simple Linear Regression Equation: y=β0+β1*x

### Predict sale of with footfall 20

```python
model.intercept_ + model.coef_*20
```

```python
model.predict([[20]])
```

## Predict sale using the trained model for new values 

```python
new_footfalls = pd.DataFrame({'Footfalls':[45,50,60]})
new_footfalls
```

```python
# Predict new sales
predicted_sales = model.predict(new_footfalls)
print("Predicted daily sales for new footfalls\n")
predicted_sales
```

```python
print(pd.DataFrame({'footfalls': new_footfalls['Footfalls'], 'PredictedSales':predicted_sales}))
```
