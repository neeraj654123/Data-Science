# MLR Evaluation Techniques

## R-square Score (Coefficient of Determination)

```python
from sklearn.metrics import r2_score

actual = [10,15,18]
pred = [12,14,16]
print("R-square:",r2_score(actual,pred))
```

```python
# So the model explains 72% of the variance in the target variable.
```

## RMSE (Root Mean Squared Error)

```python
from sklearn.metrics import mean_squared_error
import numpy as np

actual = [10,15,18]
pred = [12,14,16]

rmse = np.sqrt(mean_squared_error(actual,pred))
print("RMSE:",round(rmse,3))
```

```python
# Lower RMSE is better
```

## MAE (Mean Absolute Error)

```python
from sklearn.metrics import mean_absolute_error
import numpy as np

actual = [10,15,18]
pred = [12,14,16]

mae = mean_absolute_error(actual,pred)
print("MAE:",round(mae,2))
```

## MAPE (Mean Absolute Percentage Error)

```python
from sklearn.metrics import mean_absolute_percentage_error
import numpy as np

actual = np.array([10,15,18])
pred = np.array([12,14,16])

mape = mean_absolute_percentage_error(actual,pred)*100
print("MAPE:",round(mape,2),"%")
```
