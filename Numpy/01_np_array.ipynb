## Ndarray - The Heart of Numpy

```python
arr1 = np.array([1,2,3,4])
print("1D array:\n",arr1) 
```

```python
arr2 = np.array([[1,2,3],[4,5,6]])
print("2D array:\n", arr2)
```

#### Array of zeros, ones, range

```python
zeros = np.zeros((2,3))
zeros
```

```python
ones =np.ones((3,3))
ones
```

```python
range_array = np.arange(0,10,2)
print("Range array: ", range_array)
```

#### Array Operations

```python
array_a =  np.array([1,2,3])
array_b = np.array([10,20,30])
array_a , array_b
```

```python
# Element-wise operations
print(array_a+array_b)  # Addition
print(array_a-array_b)  # Subraction
print(array_a*array_b)  # Multiplication
print(array_b/array_a)  # Divivsion
print(array_a**2)       # Exponential
```

```python
# Broadcasting
matrix =np.array([[1,2,3],[4,5,6]])
print("Normal:\n", matrix)
print("Broadcasted:\n",matrix + 5)
```

## Indexing, Slicing, Reshaping

```python
# Indexing
mat =np.array([[10,20,30],[40,50,60]])
print(mat[1][1])
```

```python
# Slicing
print(mat[::-1,::-1])
```

```python
# Reshaping
reshaped_array = np.arange(12).reshape(3,4)
reshaped_array
```

```python
reshaped_mat = np.reshape(mat,(3,2))
reshaped_mat
```

## Useful Numpy Functions

```python
# mean
arr = np.array([4,7,3,67,2,8])
print(np.mean(arr))
```

```python
# Standard Deviation
print(np.std(arr))
```

```python
# Sort
print(np.sort(arr))
```

```python
# Linespace
lin =np.linspace(0,3,8)
print(lin)
```
