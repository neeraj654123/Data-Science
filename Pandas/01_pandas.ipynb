# Series

```python
s = pd.Series([1,2,3,4], index=['a','b','c','d'])
s
```

```python
enteries = {'swift':649000,
           'santro':391000,
           'audi':5476000,
           'elantra':1786000,
           'bolero':1091000}
car_series = pd.Series(data = enteries, name = "price")
car_series
```

```python
# Cars with price more than 10 lakhs
car_series[car_series>1000000]
```

```python
car_series[car_series>1000000].index[0]
```

```python
car_series[car_series>1000000].values[1]
```

```python
car_series[(car_series>1000000) & (car_series<2000000)]
```

#### loc[]

```python
car_series.loc['swift']
```

```python
car_series.loc[['swift','audi']]
```
